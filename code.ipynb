```python
import pandas as pd 
import numpy as np 
```

```python
dataset=pd.read_csv("heart.csv")
dataset.head()
```

```python
dataset.isnull().sum()
```

```python
dataset.info()
```

```python
dataset.head()
```

```python
dataset["ChestPainType"].unique()
```

```python
dataset["ChestPainType"]=dataset["ChestPainType"].replace({
         'ATA':0,
         'NAP':1,
         'ASY':2,
         'TA':3
})
```

```python
dataset["RestingECG"].unique()
```

```python
dataset["RestingECG"]=dataset["RestingECG"].replace({
    'Normal':0,
    'ST':1,
    'LVH':2
})
```

```python
dataset["ExerciseAngina"].unique()
```

```python
dataset["ExerciseAngina"]=dataset["ExerciseAngina"].replace({
    'N':0,
    'Y':1
})
```

```python
dataset["ST_Slope"].unique()
```

```python
dataset["ST_Slope"]=dataset["ST_Slope"].replace({
    'Up':0,
    'Flat':1,
    'Down':2
})
```

```python
dataset.head()
```

```python
dataset.info()
```

```python
x=dataset.iloc[:,[2,3,4,5,6,7,8,9,10]]
y=dataset.iloc[:,11]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
from sklearn.svm import SVC
svc=SVC()
svc.fit(x_train,y_train)
y_pred=svc.predict(x_test)
y_pred
```

```python
chestpaintype=int(input("Enter your chestpain type('ATA':0,'NAP':1,'ASY':2,'TA':3): "))
restingbp=int(input("Enter your resting BP: "))
cholesterol=int(input('Enter your cholesterol level: '))
fastingbs=int(input("Enter your fasting BS(0-1): "))
restingecg=int(input("Enter you resting ECG('Normal':0,'ST':1,'LVH':2): "))
maxhr=int(input("Enter your max HR: "))
exerciseangina=int(input("Enter your exercise angina( 'N':0,'Y':1): "))
oldpeak=int(input("Enter your old peak: "))
stslope=int(input("Enter your st slope('Up':0,'Flat':1,'Down':2): "))
new_patient=[[chestpaintype,restingbp,cholesterol,fastingbs,restingecg,maxhr,exerciseangina,oldpeak,stslope]]
new_patient_scaled=sc.transform(new_patient)
prediction=svc.predict(new_patient_scaled)
print("Here the output 0 means not heart disease but 1 means heart disease: ",prediction)
if(prediction==0):
    print("No heart disease and you are already fine!")
else:
    print("Heart issue!")
```

```python
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,roc_auc_score,confusion_matrix)
acc=accuracy_score(y_test,y_pred)
print("accuracy score= ",acc)
prec=precision_score(y_test,y_pred)
print("precision score= ",prec)
rec=recall_score(y_test,y_pred)
print("recall score= ",rec)
f1=f1_score(y_test,y_pred)
print("f1 score= ",f1)
auc=roc_auc_score(y_test,y_pred)
print("roc auc score= ",auc)
confuse=confusion_matrix(y_test,y_pred)
print("confusion matrix: \n",confuse)
```
